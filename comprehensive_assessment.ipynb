### ข้อ 1: การจัดการสตริง (String Manipulation)

กำหนดข้อมูลนักศึกษาในรูปแบบสตริง:
```
student_info = "รหัส:6601234|ชื่อ:สมชาย ใจดี|คะแนน:85"
```

ให้เขียนโปรแกรมเพื่อ:
1. แยกรหัสนักศึกษาออกมา
2. แยกชื่อนักศึกษาออกมา
3. แยกคะแนนออกมา และแปลงให้เป็น integer
4. แสดงผลลัพธ์ในรูปแบบที่อ่านง่าย

**Expected Output:**
```
รหัสนักศึกษา: 6601234
ชื่อ: สมชาย ใจดี
คะแนน: 85 (เป็น integer)
```

````python
student_info = "รหัส:6601234|ชื่อ:สมชาย ใจดี|คะแนน:85"

# พิมพ์โค้ดคำตอบที่นี่
part = student_info.split("|")
stu_id = part[0].split(":")[1]
name = part[1].split(":")[1]
score = int(part[2].split(":")[1])
print(f"รหัสนักศึกษา: {stu_id}")
print(f"ชื่อ: {name}")
print(f"คะแนน: {score}")
````

### ข้อ 2: การจัดการลิสต์ (List Operations)

กำหนดรายการสินค้าของร้านค้า:
```
products = ["นม", "ขนมปัง", "ไข่", "น้ำตาล"]
```

ให้เขียนโปรแกรมเพื่อ:
1. เพิ่ม "เนย" และ "แยม" ต่อท้ายรายการ (ใช้ append)
2. แทรก "ชา" ในตำแหน่งที่ 2 (ใช้ insert)
3. ลบ "ไข่" ออกจากรายการ (ใช้ remove)
4. แสดงรายการสินค้าทั้งหมด
5. แสดงจำนวนสินค้าทั้งหมด
6. แสดงสินค้า 3 รายการแรก

**Expected Output:**
```
รายการสินค้า: ['นม', 'ขนมปัง', 'ชา', 'น้ำตาล', 'เนย', 'แยม']
จำนวนสินค้าทั้งหมด: 6 รายการ
สินค้า 3 รายการแรก: ['นม', 'ขนมปัง', 'ชา']
```

````python
products = ["นม", "ขนมปัง", "ไข่", "น้ำตาล"]

# พิมพ์โค้ดคำตอบที่นี่
products.append("เนย")
products.append("แยม")
products.insert(2,"ชา")
products.remove("ไข่")
print(f"รายการสินค้า: {products}")
print(f"จำนวนสินค้าทั้งหมด: {len(products)}")
print(f"สินค้า 3 รายการแรก: {products[0:3]}")
````

ให้เขียนโปรแกรมเพื่อจัดการข้อมูลนักศึกษา:

2. เพิ่มนักศึกษาคนที่ 4 ลงใน dictionary
3. อัปเดตคะแนนของนักศึกษาคนใดคนหนึ่ง
4. แสดงข้อมูลนักศึกษาทั้งหมด (ให้ใช้ loop)
5. คำนวณคะแนนเฉลี่ยของนักศึกษาทั้งหมด

**Expected Output:**

````python
# พิมพ์โค้ดคำตอบที่นี่

students = {
    "6601001": {"name": "สมชาย", "score": 78},
    "6601002": {"name": "สมหญิง", "score": 85},
    "6601003": {"name": "สมหวัง", "score": 92}
}
students[6601004] = {
    "name": "สมหมาย",
    "score": 100
}
students["6601002"]["score"] = 86

print(f"นักศึกษาทั้งหมด: {len(students)}")
total_score = 0

print(f"คะแนนเฉลี่ย = {avg:.2f}")

print()
for student_id, info in students.items():
    print(f"รหัส: {student_id}")
    print(f"ชื่อ: {info['name']}")
    print(f"คะแนน: {info['score']}")
    print("-" * 20)
    total_score += info["score"]
avg = total_score / len(students)
````

### ข้อ 4: นิพจน์ทางคณิตศาสตร์ (Mathematical Expressions)

ให้เขียนโปรแกรมแปลงนิพจน์ทางคณิตศาสตร์ต่อไปนี้เป็นโค้ด Python และแสดงผลลัพธ์:

1. **พื้นที่วงกลม:** A = πr² โดยให้ r = 5
2. **นิพจน์เศษส่วน:** (15 + 25) × 3 ÷ 2⁴
3. **รากที่ 3:** ∛64 (cube root of 64)
4. **ระยะทาง:** d = √((x₂-x₁)² + (y₂-y₁)²) โดยให้ (x₁,y₁) = (3,4) และ (x₂,y₂) = (7,1)

**หมายเหตุ:** ใช้ `math` module สำหรับ pi และ sqrt

**Expected Output:**
```
1. พื้นที่วงกลม = 78.50
2. ผลลัพธ์นิพจน์ = 7.50
3. รากที่ 3 ของ 64 = 4.00
4. ระยะทาง = 5.00
```

````python
import math
 

# พิมพ์โค้ดคำตอบที่นี่
pi = math.pi
r = 5

A = pi * (r**2)
B = (15 + 25) * (3/(2**4))
C = 64 ** (1/3)
D = (((7-3)**2) + ((1-4)**2)) ** (1/2)
print(f"1. พื้นที่วงกลม = {A:.2f}")
print(f"2. ผลลัพธ์นิพจน์ = {B:.2f}")
print(f"3. รากที่ 3 ของ 64 = {C:.2f}")
print(f"4. ระยะทาง = {D:.2f}")
````

### ข้อ 5: ตัวดำเนินการเปรียบเทียบและเงื่อนไข (Comparison & Logic)

ให้เขียนโปรแกรมเพื่อ:
1. รับคะแนนสอบของนักศึกษา 3 วิชา
2. คำนวณคะแนนเฉลี่ย
3. ตรวจสอบผ่านหรือไม่ (เงื่อนไข: คะแนนเฉลี่ย >= 50 **และ** ทุกวิชา >= 40)
4. กำหนดเกรด โดยใช้เงื่อนไขต่อไปนี้:
   - A: คะแนน >= 80
   - B: คะแนน >= 70
   - C: คะแนน >= 60
   - D: คะแนน >= 50
   - F: คะแนน < 50

**ตัวอย่างการทำงาน:**
```python
# ให้รับค่าจากผู้ใช้หรือกำหนดค่าเอง
score1, score2, score3 = 75, 65, 80
```

**Expected Output (สำหรับตัวอย่างด้านบน):**
```
คะแนนวิชาที่ 1: 75
คะแนนวิชาที่ 2: 65
คะแนนวิชาที่ 3: 80
คะแนนเฉลี่ย: 73.33
ผลการสอบ: ผ่าน
เกรด: B
```

````python
# พิมพ์โค้ดคำตอบที่นี่

# สามารถรับค่าจากผู้ใช้หรือกำหนดค่าเอง
score1, score2, score3 = 75, 65, 80

print(f"คะแนนวิชาที่ 1: {score1}")

print(f"คะแนนวิชาที่ 2: {score2}")

print(f"คะแนนวิชาที่ 3: {score3}")

scores = [score1, score2, score3]
avg = (score1 + score2 + score3) / len(scores)
print(f"คะแนนเฉลี่ย: {avg:.2f}")

if avg >= 50 and min(scores)>= 40:
    print("ผลการสอบ: ผ่าน")
else:
    print("ผลการสอบ: ไม่ผ่าน:")

if avg >= 80:
    print("เกรด = A")
elif avg >= 70:
    print("เกรด = B")
elif avg >= 60:
    print("เกรด = C")
elif avg >= 50:
    print("เกรด = D")
else:
    print("เกรด = F")
````

### ข้อ 6: การวนซ้ำและการกรองข้อมูล (Loop with Filtering)

ให้เขียนโปรแกรมที่แสดงตัวเลข 1 ถึง 50 โดยใช้ loop และจัดการดังนี้:
1. ถ้าตัวเลขหารด้วย 3 ลงตัว ให้ข้ามไป (ใช้ continue)
2. ถ้าตัวเลขหารด้วย 7 ลงตัว ให้แสดง "LUCKY" แทนตัวเลข
3. กรณีอื่น ให้แสดงตัวเลขตามปกติ
4. ถ้าเจอตัวเลขที่หารด้วย 37 ลงตัว ให้หยุดการวนซ้ำ (ใช้ break)

**Expected Output (ตัวอักษรแรกๆ):**
```
1 2 4 5 LUCKY 8 10 11 13 LUCKY 16 17 19 20 22 23 25 26 LUCKY 29 31 32 34 35 ...
```

````python
# พิมพ์โค้ดคำตอบที่นี่
for i in range(0,51):
    if i % 3 == 0:
        continue
    elif i % 7 == 0:
        print("LUCKY")
    elif i % 37 == 0:
        break
    else:
        print(i)
````

### ข้อ 8: ฟังก์ชันคำนวณสถิติ (Function for Statistics)

ให้เขียนฟังก์ชัน `calculate_stats(numbers)` ที่:
1. รับ list ของตัวเลขเป็น parameter
2. คำนวณและ return dictionary ที่มี keys ต่อไปนี้:
   - `min`: ค่าที่น้อยที่สุด
   - `max`: ค่าที่มากที่สุด
   - `sum`: ผลรวมทั้งหมด
   - `count`: จำนวนตัวเลข
   - `average`: ค่าเฉลี่ย (rounded to 2 decimal places)

**ตัวอย่างการใช้:**
```python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
result = calculate_stats(scores)
print(result)
```

**Expected Output:**
```
{'min': 42, 'max': 95, 'sum': 750, 'count': 10, 'average': 75.0}
```

````python
# พิมพ์โค้ดคำตอบที่นี่
dic = {}
# ให้สร้างฟังก์ชัน calculate_stats แล้วเรียกใช้

scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
# result = calculate_stats(scores)
# print(result)
def calculate_stats(numbers):
    result = {
        "min": min(numbers),
        "max": max(numbers),
        "sum": sum(numbers),
        "count": len(numbers),
        "average": sum(numbers) / len(numbers)
    }
    return result
result = calculate_stats(scores)
print(result)
````

### ข้อ 9: ฟังก์ชันตรวจสอบรหัสผ่าน (Password Validation Function)

ให้เขียนฟังก์ชัน `check_password(password)` ที่ตรวจสอบความปลอดภัยของรหัสผ่าน:

**เงื่อนไขการตรวจสอบ:**
1. ความยาวต้องอย่างน้อย 8 ตัวอักษร
2. ต้องมีตัวเลข (0-9) อย่างน้อย 1 ตัว
3. ต้องมีตัวอักษรพิมพ์ใหญ่ (A-Z) อย่างน้อย 1 ตัว

**ฟังก์ชันจะ return:**
- `True` ถ้า password ปลอดภัย
- `False` ถ้า password ไม่ปลอดภัย พร้อมกับข้อความบอกเหตุผล

**ตัวอย่างการใช้:**
```python
check_password("Hello123")     # True, "รหัสผ่านปลอดภัย"
check_password("hello")         # False
check_password("Hello12")       # False
check_password("HELLO123")      # True
```

````python
# พิมพ์โค้ดคำตอบที่นี่
# ให้สร้างฟังก์ชัน check_password แล้วทดสอบด้วยตัวอย่างต่างๆ

# test_passwords = ["Hello123", "hello", "Hello12", "HELLO123"]
# for pwd in test_passwords:
#     check_password(pwd)
def check_password(pwd):
    if len(pwd) < 8:
        return False, "รหัสผ่านต้องมีอย่างน้อย 8 ตัวอักษร"
    if not any(char.isdigit() for char in pwd):
        return False, "ต้องมีตัวเลขอย่างน้อย 1 ตัว"
    if not any(char.isupper() for char in pwd):
        return False, "ต้องมีตัวอักษรพิมพ์ใหญ่อย่างน้อย 1 ตัว"
    return True,"รหัสผ่านปลอดภัย"
test_passwords = ["Hello123", "hello", "Hello12", "HELLO123"]
for p in test_passwords:
    print(check_password(p))
````

### ข้อ 10: โปรแกรมจัดการร้านค้า (Inventory Management System)

ให้เขียนโปรแกรมจัดการคลังสินค้าที่มี dictionary เก็บข้อมูลสินค้า และฟังก์ชันต่างๆ ดังนี้:

**โครงสร้างข้อมูล:**
- Dictionary ชื่อ `inventory` มี structure: `{"ชื่อสินค้า": {"price": ราคา, "quantity": จำนวน}}`

**ฟังก์ชันที่ต้องสร้าง:**
1. `add_product(name, price, quantity)` - เพิ่มสินค้าใหม่ไปในคลังสินค้า
2. `sell_product(name, quantity)` - ขายสินค้า (ลดจำนวน) โดยต้องตรวจสอบว่ามีจำนวนพอหรือไม่
3. `show_inventory()` - แสดงรายการสินค้าทั้งหมด พร้อม ราคา จำนวน มูลค่า
4. `total_value()` - คำนวณมูลค่าสินค้าทั้งหมด (ราคา × จำนวน)

**ตัวอย่างการใช้:**
```python
inventory = {}

add_product("นม", 25, 10)
add_product("ขนมปัง", 15, 20)
add_product("ไข่", 30, 5)

sell_product("นม", 3)      # ขายนม 3 ชิ้น

show_inventory()
print(f"มูลค่าสินค้าทั้งหมด: {total_value()} บาท")
```

**Expected Output:**
```
===== รายการสินค้า ====="
1. นม
   ราคา: 25 บาท/ชิ้น
   คงเหลือ: 7 ชิ้น
   มูลค่า: 175 บาท

2. ขนมปัง
   ราคา: 15 บาท/ชิ้น
   คงเหลือ: 20 ชิ้น
   มูลค่า: 300 บาท

3. ไข่
   ราคา: 30 บาท/ชิ้น
   คงเหลือ: 5 ชิ้น
   มูลค่า: 150 บาท

========================
มูลค่าสินค้าทั้งหมด: 625 บาท
```

````python
# พิมพ์โค้ดคำตอบที่นี่

inventory = {}
# ให้สร้างฟังก์ชัน add_product, sell_product, show_inventory, total_value
inventory = {}

def add_product(name, price, quantity):
    inventory[name] = {
        "price": price,
        "quantity": quantity
    }

def sell_product(name, quantity):
    if name in inventory and inventory[name]["quantity"] >= quantity:
        inventory[name]["quantity"] -= quantity

def show_inventory():
    print("===== รายการสินค้า =====")
    i = 1
    for name, data in inventory.items():
        price = data["price"]
        quantity = data["quantity"]
        value = price * quantity

        print(f"{i}. {name}")
        print(f"ราคา: {price} บาท/ชิ้น")
        print(f"คงเหลือ: {quantity} ชิ้น")
        print("=======================")
        print(f"มูลค่า: {value} บาท")
        print()
        i += 1

def total_value():
    total = 0
    for data in inventory.values():
        total += data["price"] * data["quantity"]
    return total

# แล้วทดสอบด้วยตัวอย่างด้านบน
# add_product("นม", 25, 10)
add_product("นม", 25, 10)
# add_product("ขนมปัง", 15, 20)
add_product("ขนมปัง", 15, 20)
# add_product("ไข่", 30, 5)
add_product("ไข่", 30, 5)
# sell_product("นม", 3)ห
sell_product("นม", 3)
# show_inventory()
show_inventory()
# print(f"มูลค่าสินค้าทั้งหมด: {total_value()} บาท")
print(f"มูลค่าสินค้าทั้งหมด: {total_value()} บาท")
````
