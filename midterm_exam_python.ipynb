### ข้อ 3. จงเขียนโปรแกรมคำนวณพื้นที่สี่เหลี่ยมผืนผ้า โดยรับค่าความกว้างและความยาวจากผู้ใช้ (2 คะแนน)

*ตัวอย่างการทำงาน:*
```
ป้อนความกว้าง: 5
ป้อนความยาว: 8

ผลลัพธ์:
พื้นที่สี่เหลี่ยมผืนผ้า = 40.00 ตารางหน่วย
```

````python
# พิมพ์โค้ดคำตอบที่นี่
width = float(input("ป้อนความกว้าง : "))
length = float(input("ป้อนความยาว : "))

area = width * length

print(f"ผลลัพธ์ : พื้นที่สี่เหลี่ยมผืนผ้า = {area:.2f} ตารางหน่วย")
````

### ข้อ 4. กำหนดให้ List ของคะแนนสอบนักศึกษา ดังนี้ (2 คะแนน)

```python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
```

จงเขียนโปรแกรมเพื่อ:
- a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)

````python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]

# a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)

average = sum(scores) / len(scores)

print(f"คะแนนเฉลี่ย = {average:.2f}")
````

### ข้อ 5. กำหนดให้ List ของชื่อนักศึกษา ดังนี้ (2 คะแนน)

```python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]
```

จงเขียนโปรแกรมเพื่อ:
- a) เพิ่มชื่อ "สมพร" ต่อท้าย List
- b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
- c) ลบชื่อ "สมศรี" ออกจาก List
- d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด

````python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]

# a) เพิ่มชื่อ "สมพร" ต่อท้าย List
students.append("สมพร")
print(students)
# b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
students.insert(2,"สมใจ")
print(students)
# c) ลบชื่อ "สมศรี" ออกจาก List
students.remove("สมศรี")
print(students)
# d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด
print(students)
````

### ข้อ 14. จงเขียนโปรแกรม FizzBuzz โดยแสดงเลข 1-30 ตามเงื่อนไขดังนี้ (2 คะแนน)

- ถ้าหารด้วย 3 ลงตัว ให้พิมพ์ "Fizz"
- ถ้าหารด้วย 5 ลงตัว ให้พิมพ์ "Buzz"
- ถ้าหารด้วย 3 และ 5 ลงตัว ให้พิมพ์ "FizzBuzz"
- กรณีอื่นๆ ให้พิมพ์ตัวเลขปกติ

*ตัวอย่างผลลัพธ์: 1, 2, Fizz, 4, Buzz, Fizz, 7, 8, Fizz, Buzz, 11, Fizz, 13, 14, FizzBuzz, ...*

````python
# พิมพ์โค้ดคำตอบที่นี่
for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````
